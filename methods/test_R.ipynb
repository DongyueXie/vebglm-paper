```python
torch.manual_seed(1)
n = 500
p = 100
X = torch.randn(n,p)
s = 2
b = torch.cat((torch.ones(s)*2,torch.zeros(p-s)))
b = b.type(torch.FloatTensor)
b = b.reshape([p,1])
mu = torch.matmul(X,b)
mu = mu.squeeze()
prob = torch.exp(mu)/(1+torch.exp(mu))
y = torch.bernoulli(prob)
print(torch.bincount(y.int()))

X_test = torch.randn(3000,p)
mu_test = torch.matmul(X_test,b).squeeze()
prob_test = torch.exp(mu_test)/(1+torch.exp(mu_test))
y_test = torch.bernoulli(prob_test)
```

```python
from bayesregR import bayesregR
model = bayesregR(family='gaussian')
model.fit(X=X,y=y)
print(model.run_time)
print(model.predict_prob(X_test))
print(model.predict(X_test))
print(roc_auc_score(y_test, model.predict_prob(X_test)))
```
