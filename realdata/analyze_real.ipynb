```python
import pickle 
import pandas as pd
import sys
import os
current_dir = os.getcwd()
parent_dir = os.path.dirname(current_dir)
sys.path.append(parent_dir)
parent_dir = os.path.dirname(parent_dir)
sys.path.append(f'{parent_dir}/VEBGLM/src')
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
import pickle

import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
datasets_name = ['adult', 'annealing', 'Ionosphere', 'heart_disease', 'abalone', 'Pediatric']
```

```python

pd.set_option('display.max_rows', None)
res_all = []
for data_name in datasets_name:
    with open(f'{current_dir}/results/{data_name}-final.pkl', 'rb') as f:
        res = pickle.load(f)
    res_all = res_all + res


results_df = pd.DataFrame(res_all)
```

```python
results_df = results_df.drop(['data_name'], axis=1)
```

```python
res = results_df.groupby(['model', 'metric']).mean().reset_index()
```
